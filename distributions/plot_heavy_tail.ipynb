```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma


# ----------------------------
# Log PDF definitions
# ----------------------------
def log_student_t(x, nu):
    """Log PDF of Student-t with df=nu"""
    c = gamma((nu + 1) / 2) / (np.sqrt(nu * np.pi) * gamma(nu / 2))
    return np.log(c) - (nu + 1) / 2 * np.log(1 + x**2 / nu)


def log_gen_gaussian(x, beta, alpha=np.sqrt(2.0)):
    """Log PDF of Generalized Gaussian"""
    c = beta / (2 * alpha * gamma(1 / beta))
    return np.log(c) - (np.abs(x) / alpha) ** beta


def log_gaussian(x):
    """Log PDF of standard Gaussian"""
    return -0.5 * np.log(2 * np.pi) - 0.5 * x**2


# ----------------------------
# Grid
# ----------------------------
x = np.linspace(-5, 5, 500)

# Parameters
nu_list = [0.1, 1.0, 10.0, 100.0]
beta_list = [0.4, 0.8, 1.0, 1.2, 1.6]

# ----------------------------
# Plot
# ----------------------------
# colors = plt.get_cmap("magma")(np.linspace(0, 1, 4))
colors = plt.get_cmap("Dark2").colors

plt.rcParams.update({"font.size": 16, "mathtext.fontset": "cm", "font.family": "serif"})

fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True)

# ---- Left: Student-t ----
ax = axes[0]
for nu, c in zip(nu_list, colors):
    ax.plot(x, log_student_t(x, nu), color=c, lw=3, label=rf"$\nu \,=\, {nu}$")

# Gaussian reference
ax.plot(x, log_gaussian(x), "--", lw=3, color="tab:brown", label=r"$\mathcal{N}$")

ax.set_title("Student t", fontsize=26)
ax.set_ylabel(r"$\log[p(\mathcal{X}(x))]$", fontsize=26)

ax.grid(True, alpha=0.6)
ax.legend(loc="lower center", ncol=2, frameon=True, fontsize=18)


# ---- Right: Generalized Gaussian ----
ax = axes[1]
for beta, c in zip(beta_list, colors):
    ax.plot(x, log_gen_gaussian(x, beta), color=c, lw=3, label=rf"$\beta \,=\, {beta}$")

# Gaussian reference
ax.plot(x, log_gaussian(x), "--", lw=3, color="tab:brown", label=r"$\mathcal{N}$")

ax.set_title("Leptokurtic", fontsize=26)

ax.grid(True, alpha=0.6)
ax.legend(loc="lower center", ncol=2, frameon=True, fontsize=18)

# Layout
plt.tight_layout()
plt.show()
```
