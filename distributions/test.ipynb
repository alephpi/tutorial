```python
import torch
from torch.distributions.utils import logits_to_probs


class ApproxBernoulli(torch.distributions.RelaxedBernoulli):
    def rsample(self, sample_shape=torch.Size()):  # noqa
        x: torch.Tensor = super().rsample(sample_shape)
        return x - x.detach() + (x > 0.5).to(x.dtype)


b = 16
n = 6
t = 100
tau = 1
```

```python
w_logits = torch.randn(b, n, t) * 100
qw = ApproxBernoulli(logits=w_logits, temperature=torch.full_like(w_logits, tau))
qw_infer = logits_to_probs(w_logits, is_binary=True)
```

```python
torch.all(qw_infer == qw.probs)
```
