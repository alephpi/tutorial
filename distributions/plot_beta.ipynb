```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# 定义 x 取值范围（Beta 分布定义在 [0,1]）
x = np.linspace(0, 1, 500)

# 多组参数 (α, β)
params = [(0.8, 0.8), (2, 0.8), (0.8, 2), (3, 2), (2, 3)]

# 绘图
plt.figure(figsize=(8, 6))
plt.xlim(0, 1)
plt.ylim(0, 2.5)

for a, b in params:
    y = beta.pdf(x, a, b)
    plt.plot(x, y, label=f"α={a}, β={b}")

# 图像格式设置
plt.title("Beta Distribution", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("PDF", fontsize=12)
plt.legend()
plt.grid(True)

plt.show()
```

```python
from itertools import product
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# x 定义域
x = np.linspace(0, 1, 500)

# m 与 lambda 取值
m_values = [0.3, 0.5, 0.7]
lambda_values = [4, 10]
combination = list(product(m_values, lambda_values)) + [(0.5, 0.0)]
combination = sorted(combination, key=lambda x: (x[1], x[0]))

# 创建画布
plt.figure(figsize=(8, 6))
plt.xlim(0, 1)
plt.ylim(0, 3.0)

# 遍历所有组合
for m, lam in combination:
    # 重参数化 -> alpha, beta
    alpha = 1 + m * lam
    beta_param = 1 + (1 - m) * lam

    # PDF
    y = beta.pdf(x, alpha, beta_param)

    # 绘制曲线
    plt.plot(x, y, label=f"m={m}, λ={lam}")

# 图像设置
plt.title("Beta Distribution with PERT parametrization", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("PDF", fontsize=12)
plt.grid(True)

# 图例较多，缩小字体
plt.legend(fontsize=8, ncol=1)

plt.show()
```
