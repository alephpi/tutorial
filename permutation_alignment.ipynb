```python
A = np.random.random((M, N))
```

```python
# permutation
import numpy as np


def random_permutation_matrix(n, scale=False):
    """
    Create a random n by n permutation matrix
    """
    perm = np.random.permutation(n)
    if scale:
        P = (np.eye(n, dtype=int) * np.diag(np.random.random(n)))[perm]
    else:
        P = (np.eye(n, dtype=int))[perm]
    return P


# Create a random 4×4 permutation matrix
P = random_permutation_matrix(3)
```

```python
X = A @ S
```
