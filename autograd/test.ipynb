```python
import torch

x1 = torch.tensor(1.0, requires_grad=True)
x2 = torch.tensor(2.0, requires_grad=True)

a = x1 * x2  # Mul
y1 = torch.log(a)  # Log
y2 = torch.sin(x2)  # Sin
w = y1 * y2  # Mul
z = w  # 输出
```
