```python
# exponential sum of exp(2i*pi*x)

import numpy as np


def exp_sum(n, M=10):
    """
    Compute the exponential sum of exp(2i*pi*x) for i=0 to M-1.
    """
    return np.sum(np.exp(2j * np.pi * n * np.arange(M))) / M
```
