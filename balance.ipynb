```python
import pandas as pd
```

```python
# import csv, first column as datetime, second as float
df = pd.read_csv("balance.csv", sep=",", parse_dates=[0], decimal=",", thousands=" ")
```

```python
df
```

```python
df["year_month"] = df["dateOp"].dt.to_period("M")
```

```python
expenses_df = df[df["amount"] < 0]
```

```python
monthly_expenses = expenses_df.groupby("year_month")["amount"].sum().reset_index()
```
