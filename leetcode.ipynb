## 最小无重复字串

```python
s = "abcabcbb"
s_ = "pwwkew"
```

```python
def lengthOfLongestSubstring(s: str) -> int:
    occ = set()
    n = len(s)
    l, r, ans = 0, 0, 0
    substring = ""
    occ.add(s[l])
    while l < n:
        while r + 1 < n and s[r + 1] not in occ:
            occ.add(s[r + 1])
            r += 1
        ans = max(ans, r - l + 1)
        if len(s[l : r + 1]) > len(substring):
            substring = s[l : r + 1]
        occ.remove(s[l])
        l += 1
    print(s[l : r + 1])
    return ans, substring
```

```python
lengthOfLongestSubstring(s)
```

```python
lengthOfLongestSubstring(s_)
```
