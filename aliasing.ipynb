# space aliasing

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive


def plot_images(panel):
    # 设置图像大小为10000x10000像素
    image_size = 1000

    # 定义采样率 (每像素对应的空间单位)
    sampling_rate = 10**panel  # 采样率，控制白色方块的间隔

    # 创建空白图像
    image = np.zeros((image_size, image_size))

    # 计算采样步长
    step_size = int(1 / sampling_rate)  # 采样步长，与采样率成反比

    # 绘制斜线的白色方块
    for x in range(0, image_size, step_size):
        y = -x + image_size  # 对应于 y = -x + 1 的斜线在离散像素空间中的映射
        if 0 <= y < image_size:
            image[
                y - step_size // 2 : y + step_size // 2 + 1,
                x - step_size // 2 : x + step_size // 2 + 1,
            ] = 1  # 设置为白色（亮度为1）

    # 计算二维傅里叶变换
    fft_image = np.fft.fftshift(np.fft.fft2(image))
    magnitude_spectrum = np.log(np.abs(fft_image) + 1)  # 取对数以便可视化

    # 可视化原图像和傅里叶频谱
    fig, ax = plt.subplots(1, 2, figsize=(12, 6))

    # 原图像
    ax[0].imshow(image, cmap="gray")
    ax[0].set_title("Original Line Segment Image (Sampled)")
    ax[0].set_xlabel("x")
    ax[0].set_ylabel("y")

    # 傅里叶频谱
    ax[1].imshow(magnitude_spectrum, cmap="gray")
    ax[1].set_title("2D Fourier Transform (Magnitude Spectrum)")
    ax[1].set_xlabel("Frequency (u)")
    ax[1].set_ylabel("Frequency (v)")

    plt.show()


interactive_plot = interactive(plot_images, panel=(-2, 0, 0.1))
output = interactive_plot.children[-1]
interactive_plot
```
