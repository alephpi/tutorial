```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
```

```python
t = np.linspace(-2 * np.pi, 2 * np.pi, 1000)
freq = 10 * np.sin(t) + 30
plt.plot(freq)
```

```python
signal = np.sin(freq * t)
plt.plot(signal[200:600])
```

```python
signal_transformed = sp.signal.hilbert(signal)
```

```python
import cmath

angle = np.vectorize(cmath.phase)(signal_transformed)
plt.plot(angle[200:600])
```
